```python
import cv2
import os
import numpy as np

# ভিডিওর পাথ
video_path = 'inputs/my_video/source.mp4'
# মাস্ক সেভ করার ফোল্ডার
mask_out_dir = 'inputs/my_video_mask_frames/'
os.makedirs(mask_out_dir, exist_ok=True)

cap = cv2.VideoCapture(video_path)
fps = int(cap.get(cv2.CAP_PROP_FPS))
frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

print(f"Total frames: {frame_count}")

count = 0
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # একটি কালো মাস্ক ইমেজ তৈরি করা (ভিডিওর সাইজের সমান)
    mask = np.zeros(frame.shape[:2], dtype=np.uint8)

    # ধূসর স্কেল (Grayscale) এ কনভার্ট করা
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # সাদা টেক্সট খোঁজার জন্য থ্রেশহোল্ড (Threshold) সেট করা
    # Hoofoot.net সাধারণত সাদা রঙের হয়
    _, thresh = cv2.threshold(gray, 200, 255, cv2.THRESH_BINARY)

    # কনট্যুর (আকৃতি) খুঁজে বের করা
    contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    for cnt in contours:
        x, y, w, h = cv2.boundingRect(cnt)
        aspect_ratio = w / float(h)
        area = cv2.contourArea(cnt)

        # শুধুমাত্র নির্দিষ্ট সাইজের টেক্সট ফিল্টার করা (যাতে অন্য সাদা জিনিস বাদ পড়ে)
        # আপনি এখানে x, y, w, h এর লিমিট বাড়াতে/কমাতে পারেন
        if 80 < w < 400 and 10 < h < 60 and aspect_ratio > 2.0:
            # টেক্সটের চারপাশে একটু বাড়তি জায়গা (Padding) রাখা
            pad = 10
            cv2.rectangle(mask, (x-pad, y-pad), (x+w+pad, y+h+pad), 255, -1)

    # মাস্ক ফ্রেম সেভ করা
    mask_filename = os.path.join(mask_out_dir, f"{count:05d}.png")
    cv2.imwrite(mask_filename, mask)

    count += 1
    if count % 100 == 0:
        print(f"Processed {count} frames...")

cap.release()
print("✅ Auto-Masking complete! Saved to:", mask_out_dir)
```
